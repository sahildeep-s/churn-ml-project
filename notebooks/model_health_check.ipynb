```python
import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score
from src.train_model import load_data, train_test_split_xy, build_preprocessor, build_models

print("Loading data...")
df = load_data()

print("Splitting...")
X_train, X_test, y_train, y_test = train_test_split_xy(df)

print("Training...")
models = build_models(build_preprocessor())
results = {}

for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)
    proba = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, proba)
    results[name] = auc
    print(f"{name}: AUC = {auc:.3f}")

print("\nModel Health: " + ("✅ GOOD" if all(v > 0.8 for v in results.values()) else "⚠️ INVESTIGATE"))
```
